## Question 16 — Student Marks Processor

**Problem:**  
A teacher has a list of marks of all students in a class and wants key statistics.

**Requirements:**
- Create a function that accepts a **list of student marks**.
- Using a loop (or appropriate logic), calculate and **return**:
  - Highest mark
  - Lowest mark
  - Total marks
  - Average marks
- Display the results outside the function.

**Focus:** Function + list + loop + `return`

````python
def marks_processor(marks):
    if len(marks) == 0:
        return None, None, 0, 0

    total = 0
    highest = marks[0]
    lowest = marks[0]

    for mark in marks:
        total += mark
        if mark > highest:
            highest = mark
        if mark < lowest:
            lowest = mark

    average = total / len(marks)
    return highest, lowest, total, average

marks = [78, 85, 92, 66, 88]
highest, lowest, total, average = marks_processor(marks)
print("Highest:", highest)
print("Lowest:", lowest)
print("Total:", total)
print("Average:", average)
````

## Question 24 — Nested Order Processing

**Problem:**  
An order processing system calculates subtotal, discount, tax, and final amount. Organize this using nested functions.

**Requirements:**
- Create an outer function called `process_order()`.
- Inside it, create nested functions for:
  - Calculate subtotal
  - Calculate discount
  - Calculate tax
  - Calculate final amount
- Return the final order amount from the outer function.
- Demonstrate with sample values.

**Focus:** Nested functions + arguments + `return` + real-world logic

````python
def process_order(price, quantity, discount_rate, tax_rate):
    def calculate_subtotal():
        return price * quantity

    def calculate_discount(subtotal):
        return subtotal * discount_rate / 100

    def calculate_tax(amount):
        return amount * tax_rate / 100

    def calculate_final():
        subtotal = calculate_subtotal()
        after_discount = subtotal - calculate_discount(subtotal)
        return after_discount + calculate_tax(after_discount)

    return calculate_final()

final_amount = process_order(1500, 3, 10, 5)
print("Final order amount: Rs.", final_amount)
````

**Problem:**  
Build a shopping billing system that can handle multiple products.

**Requirements:**

````python
def product_total(price, quantity):
    return price * quantity

def calculate_subtotal(*totals):
    return sum(totals)

def calculate_discount(subtotal, rate=10):
    return subtotal * rate / 100

def calculate_tax(amount, rate=5):
    return amount * rate / 100

products = [
    {"name": "Keyboard", "price": 2500, "quantity": 1},
    {"name": "Mouse", "price": 1200, "quantity": 2},
    {"name": "USB", "price": 800, "quantity": 1}
]

individual_totals = []
print("---------- SHOPPING BILL ----------")
for product in products:
    total = product_total(product["price"], product["quantity"])
    individual_totals.append(total)
    print(f'{product["name"]}: {product["price"]} x {product["quantity"]} = Rs. {total}')

subtotal = calculate_subtotal(*individual_totals)
discount = calculate_discount(subtotal)
after_discount = subtotal - discount
tax = calculate_tax(after_discount)
final_bill = after_discount + tax

print("-----------------------------------")
print("Subtotal: Rs.", subtotal)
print("Discount: Rs.", discount)
print("Tax: Rs.", tax)
print("Final bill: Rs.", final_bill)
````

## Question 29 — Student Result Management System

**Problem:**  
Create a complete student result system using functions.

**The system should:**
1. Accept student information
2. Accept multiple subject marks
3. Calculate total
4. Calculate percentage
5. Determine grade
6. Determine pass/fail
7. Display a final result card

**Requirements:**
- Use **multiple functions**
- Use arguments and `return`
- Use lists
- Use conditions and loops
- Use at least one advanced concept (`*args`, `**kwargs`, nested functions, or lambda)
- Do **not** write the entire program inside one function

**Focus:** Combining multiple function concepts

````python
def get_student_info(name, roll_no):
    return {"Name": name, "Roll No": roll_no}

def calculate_total(marks):
    total = 0
    for mark in marks:
        total += mark
    return total

def calculate_percentage(total, subjects):
    maximum_marks = subjects * 100
    return (total / maximum_marks) * 100 if maximum_marks else 0

def get_grade(percentage):
    if percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"

def pass_status(marks):
    return "Pass" if all(mark >= 40 for mark in marks) else "Fail"

marks = [78, 85, 69, 74]
student = get_student_info("Ahmed", "STU-01")
total = calculate_total(marks)
percentage = calculate_percentage(total, len(marks))
grade = get_grade(percentage)
status = pass_status(marks)

print("======= STUDENT RESULT CARD =======")
for key, value in student.items():
    print(f"{key}: {value}")
for index, mark in enumerate(marks, start=1):
    print(f"Subject {index}: {mark}")
print("Total marks:", total)
print("Percentage:", round(percentage, 2), "%")
print("Grade:", grade)
print("Result:", status)
````

````python
def customer_info(name, city):
    return {"Customer": name, "City": city}

def get_products():
    return [
        {"name": "Headphones", "price": 3000, "quantity": 1},
        {"name": "Keyboard", "price": 2500, "quantity": 2},
        {"name": "Mouse", "price": 1000, "quantity": 1}
    ]

def calculate_item_total(price, quantity):
    return price * quantity

def calculate_subtotal(*item_totals):
    return sum(item_totals)

def apply_discount(subtotal, rate=10):
    return subtotal * rate / 100

def calculate_tax(amount, rate=5):
    return amount * rate / 100

def delivery_charge(subtotal):
    if subtotal >= 5000:
        return 0
    return 250

def make_order_summary(customer, products, discount_rate=10, tax_rate=5):
    # Nested function to format each product line.
    def product_line(product):
        total = calculate_item_total(product["price"], product["quantity"])
        return total

    item_totals = []
    print("========== ORDER SUMMARY ==========")
    for product in products:
        item_total = product_line(product)
        item_totals.append(item_total)
        print(f'{product["name"]} - {product["quantity"]} x Rs. {product["price"]} = Rs. {item_total}')

    subtotal = calculate_subtotal(*item_totals)
    discount = apply_discount(subtotal, discount_rate)
    discounted_total = subtotal - discount
    tax = calculate_tax(discounted_total, tax_rate)
    delivery = delivery_charge(subtotal)
    final_amount = discounted_total + tax + delivery

    print("-----------------------------------")
    for key, value in customer.items():
        print(f"{key}: {value}")
    print("Subtotal: Rs.", subtotal)
    print("Discount: Rs.", discount)
    print("Tax: Rs.", tax)
    print("Delivery: Rs.", delivery)
    print("Final amount: Rs.", round(final_amount, 2))
    return final_amount

customer = customer_info("Ahmed", "Karachi")
products = get_products()
# Lambda is used to quickly create a short product-name list.
product_names = list(map(lambda item: item["name"], products))
print("Products in order:", ", ".join(product_names))
make_order_summary(customer, products)
````
